```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#put your x and y values here
x = np.array(['here']).reshape(-1, 1)
y = np.array(['and here'])
```

```python
plt.plot(x, y, '+')
plt.show()
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
a = model.coef_[0]
b = model.intercept_
```

```python
print(a , b)
```

```python
#here you can predict the value of y for a given x value
plt.plot(x, y, '+')
plt.plot(x , model.predict(x))
plt.show()
```

```python
result = model.predict([['put your x value here to predict y']])
print(result)
```

```python
#here is your y=mx+b
m = a.round(2)
b = b.round(2)
print(f"The equation of the line is: y = {m}x + {b}")
```
